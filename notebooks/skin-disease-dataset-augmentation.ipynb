```python
import os
import cv2
import numpy as np
from tqdm import tqdm
import albumentations as A
from albumentations.pytorch import ToTensorV2
```

```python
INPUT_DIR = "/kaggle/input/datasets/ismailpromus/skin-diseases-image-dataset/IMG_CLASSES" 
OUTPUT_DIR = "/kaggle/working/augmented_dataset"

if not os.path.exists(OUTPUT_DIR):
    os.makedirs(OUTPUT_DIR)
```

```python
train_transform = A.Compose([
    A.Resize(224, 224),
    A.HorizontalFlip(p=0.5),
    A.VerticalFlip(p=0.2),
    A.Rotate(limit=15, p=0.5),
    A.RandomBrightnessContrast(p=0.5),
    A.HueSaturationValue(p=0.5),
])
```

```python
for class_name in os.listdir(INPUT_DIR):
    class_input_path = os.path.join(INPUT_DIR, class_name)
    class_output_path = os.path.join(OUTPUT_DIR, class_name)
    
    if not os.path.exists(class_output_path):
        os.makedirs(class_output_path)
    
    for img_name in tqdm(os.listdir(class_input_path), desc=f"Processing {class_name}"):
        img_path = os.path.join(class_input_path, img_name)
        img = cv2.imread(img_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        
        # Save original
        cv2.imwrite(os.path.join(class_output_path, img_name), cv2.cvtColor(img, cv2.COLOR_RGB2BGR))
        
        # Apply augmentation 4 times per image
        for i in range(4):
            augmented = train_transform(image=img)['image']
            aug_name = img_name.split('.')[0] + f"_aug{i}.jpg"
            cv2.imwrite(os.path.join(class_output_path, aug_name), cv2.cvtColor(augmented, cv2.COLOR_RGB2BGR))
```
